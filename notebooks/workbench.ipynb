# Локальная проверка переносимого DevOps-шаблона

Этот ноутбук только читает локальные файлы. Он не обращается к серверу, не выполняет SSH, OpenTofu apply, Ansible и операции сброса. Настройка и запуск описаны в `../README.md`.


```python
from pathlib import Path
import json

root = next(path for path in (Path.cwd(), *Path.cwd().parents) if (path / 'cpp/CMakeLists.txt').is_file())
print('Проект:', root.name)
```

## Проверка структур JSON

Чтение демо-варианта, проверка синтаксиса и подсчёт VM, сетей и workload. Никакие секреты не загружаются.


```python
demo = root / 'configs/source/demo'
configurations = sorted(demo.rglob('*.json'))
for path in configurations:
    json.loads(path.read_text(encoding='utf-8'))
workloads = sorted((root / 'configs/source/workloads/demo').rglob('*.json'))
for path in workloads:
    json.loads(path.read_text(encoding='utf-8'))
print({'vm': sum(p.name.endswith('.vm.json') for p in configurations), 'networks': sum(p.name.endswith('.network.json') for p in configurations), 'workloads': len(workloads)})
```
